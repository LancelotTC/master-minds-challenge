```python
import numpy as np
import matplotlib.pyplot as plt

days = np.array([1, 2, 3, 4, 5, 6, 7])
day_names = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

days_sin = np.sin(2 * np.pi * days / 7)
days_cos = np.cos(2 * np.pi * days / 7)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(days, np.zeros_like(days), '-o', color='#1f77b4', markersize=10)

for i, txt in enumerate(day_names):
    ax1.annotate(txt, (days[i], 0.05), ha='center', fontsize=12)

ax1.set_title("Linear Encoding\n(Mon and Sun are far apart)", fontsize=14, pad=20)
ax1.set_yticks([])
ax1.set_xticks(days)
ax1.set_xlim(0.5, 7.5)
ax1.set_ylim(-0.2, 0.3)
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax1.spines['left'].set_visible(False)

ax1.annotate('', xy=(1, -0.05), xytext=(7, -0.05),
            arrowprops=dict(edgecolor='red', arrowstyle='<->', linestyle='--', connectionstyle="arc3,rad=0.2"))
ax1.text(4, -0.15, 'Large Distance', color='red', ha='center', fontsize=11)


ax2.plot(days_sin, days_cos, 'o', color='#ff7f0e', markersize=10)

circle_angles = np.linspace(0, 2 * np.pi, 100)
ax2.plot(np.sin(circle_angles), np.cos(circle_angles), '--', color='gray', alpha=0.3)

for i, txt in enumerate(day_names):
    ax2.annotate(txt, (days_sin[i]*1.2, days_cos[i]*1.2), ha='center', va='center', fontsize=12)


ax2.annotate('', xy=(days_sin[0]*0.9, days_cos[0]*0.9), xytext=(days_sin[-1]*0.9, days_cos[-1]*0.9),
            arrowprops=dict(edgecolor='red', arrowstyle='<->', linestyle='--', connectionstyle="arc3,rad=0.2"))
ax2.text(days_sin[0]*0.4, days_cos[0]*1.1, 'Small Distance', color='red', ha='center', fontsize=11)


ax2.set_title("Cyclical Encoding (Sin / Cos)\n(Mon and Sun are neighbors)", fontsize=14, pad=20)
ax2.set_xlabel("Sin Feature", fontsize=12)
ax2.set_ylabel("Cos Feature", fontsize=12)
ax2.set_xlim(-1.5, 1.5)
ax2.set_ylim(-1.5, 1.5)
ax2.set_aspect('equal')
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

days = ['Wed', 'Thu', 'Fri', 'Sat\n(Off)', 'Sun\n(Off)', 'Mon', 'Tue']
is_off = [0, 0, 0, 1, 1, 0, 0]

days_until_off = [3, 2, 1, 0, 0, 4, 3]
days_until_work = [0, 0, 0, 2, 1, 0, 0]

x = np.arange(len(days))

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(x, days_until_off, marker='o', linewidth=3, markersize=10, color='#1f77b4', label='Days Until Next Day Off')
ax.plot(x, days_until_work, marker='s', linewidth=3, markersize=10, color='#ff7f0e', label='Days Until Next Workday')

ax.axvspan(2.5, 4.5, color='gray', alpha=0.15, label='Weekend / Holiday')

ax.set_title("How Countdown Features React to Time", fontsize=16, pad=20, weight='bold')
ax.set_xticks(x)
ax.set_xticklabels(days, fontsize=12)
ax.set_yticks(range(0, 5))
ax.set_ylabel("Number of Days", fontsize=12)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', linestyle='--', alpha=0.7)

ax.legend(fontsize=11, loc='upper left')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
fig, ax = plt.subplots(figsize=(12, 4))
ax.axis('off')

levels = ["Level 0", "Level 1", "Level 2", "Level 3"]
titles = ["SAFE", "CAUTION", "DANGERS EXIST", "DO NOT FLY"]
desc = [
    "Normal Operations.\nMinimum Risk.",
    "Heightened Monitoring.\nMinor Reroutes possible.",
    "Active Military Activity.\nReroutes Required.",
    "Airspace Closed.\nSevere Risk to Life."
]

colors = ['#d4edda', '#fff3cd', '#ffe5b4', '#f8d7da']
edge_colors = ['#c3e6cb', '#ffeeba', '#ffc071', '#f5c6cb']
text_colors = ['#155724', '#856404', '#cc5500', '#721c24']

width_per_level = 3
rect_height = 2.5
rect_y = 0.5

for i in range(4):
    x = i * width_per_level
    rect = patches.Rectangle((x, rect_y), width_per_level, rect_height, 
                             facecolor=colors[i], edgecolor=edge_colors[i], linewidth=2, zorder=1)
    ax.add_patch(rect)
    
    ax.text(x + (width_per_level/2), rect_y + 2.1, levels[i], 
            ha='center', va='center', fontsize=12, weight='bold', color=text_colors[i], zorder=2)
    
    ax.text(x + (width_per_level/2), rect_y + 1.6, titles[i], 
            ha='center', va='center', fontsize=16, weight='bold', color=text_colors[i], zorder=2)
    
    ax.text(x + (width_per_level/2), rect_y + 0.6, desc[i], 
            ha='center', va='center', fontsize=11, color='#333333', linespacing=1.5, zorder=2)


needle_origin_x = 0.5
needle_tip_x = 10.5
ax.annotate('', xy=(needle_tip_x, rect_y + 0.2), xytext=(needle_origin_x, rect_y + 0.2), 
            arrowprops=dict(facecolor='#404040', edgecolor='none', width=3, headwidth=15, zorder=3))

ax.text(needle_tip_x, rect_y - 0.2, "Increasing Passengers Risk Aversion", 
        ha='right', va='top', fontsize=11, color='#404040', weight='bold', style='italic')

ax.text(6, 3.7, "Conflict Features: Operational Severity Levels", 
        ha='center', fontsize=16, weight='bold', color='#2c3e50')

ax.set_xlim(-0.5, 12.5)
ax.set_ylim(-0.5, 4.2)

plt.tight_layout()

plt.show()
```
